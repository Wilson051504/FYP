```python

import pandas as pd

# STEP 1: Load the dataset
file_path = "/content/data.csv"

df = pd.read_csv(
    file_path,
    encoding="cp1252",
    low_memory=False
)

print("Original dataset size:", df.shape)

# STEP 2: Filter Malaysia
if "economycode" in df.columns:
    malaysia = df[
        df["economycode"].astype(str).str.upper() == "MYS"
    ].copy()
elif "economy" in df.columns:
    malaysia = df[
        df["economy"].astype(str).str.strip().str.casefold()
        == "malaysia"
    ].copy()
else:
    raise ValueError(
        "Cannot find the country/economy column. "
        "Please check the column names."
    )

print("Malaysia sample size:", malaysia.shape)

# STEP 3: Check required variables
variables = [
    "fin24", "fin24a",
    "female", "age", "educ", "inc_q", "emp_in"
]

print("\nVariable availability:")
for col in variables:
    print(col, "->", col in malaysia.columns)

# STEP 4: Check missing values and unique values
existing = [
    col for col in variables
    if col in malaysia.columns
]

print("\nMissing values:")
print(malaysia[existing].isnull().sum())

print("\nNumber of unique values:")
print(malaysia[existing].nunique(dropna=True))

# STEP 5: Inspect target candidates
for col in ["fin24", "fin24a"]:
    if col in malaysia.columns:
        print(f"\nDistribution of {col}:")
        print(
            malaysia[col]
            .value_counts(dropna=False)
            .sort_index()
        )

# STEP 6: Inspect predictor coding
for col in ["female", "age", "educ", "inc_q", "emp_in"]:
    if col in malaysia.columns:
        print(f"\nDistribution of {col}:")
        print(
            malaysia[col]
            .value_counts(dropna=False)
            .sort_index()
        )
```

```python
import pandas as pd

# 1. 读取数据
file_path = "data.csv"  # 若在 Colab 上请放 "/content/data.csv"
df = pd.read_csv(file_path, encoding="cp1252", low_memory=False)
print("Global dataset shape:", df.shape)

# 2. 筛选马来西亚样本
malaysia = df[df["economycode"].astype(str).str.upper() == "MYS"].copy()
print("Malaysia sample size:", malaysia.shape)

# 3. 检查核心自变量与目标变量
variables = ["fin24", "fin24a", "female", "age", "educ", "inc_q", "emp_in"]

print("\n--- Missing Values in Malaysia Sample ---")
print(malaysia[variables].isnull().sum())

print("\n--- Value Counts of Predictors ---")
for col in ["female", "educ", "inc_q", "emp_in"]:
    print(f"\n{col}:\n", malaysia[col].value_counts(dropna=False).sort_index())

print("\nAge summary:")
print(malaysia["age"].describe())
```
